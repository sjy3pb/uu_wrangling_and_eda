```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_csv("data/ForeignGifts_edu.csv")

df.head()
```

First, lets create a data frame 'uva' that only contains gifts to UVA

```python
uva = df[df['Institution Name'] == 'University of Virginia']

uva.head()
```

We can group by country and calculate the sums to show which countries are currently giving the most to UVA

```python
country_uva_total_dons = uva.groupby('Country of Giftor')['Foreign Gift Amount'].sum().sort_values(ascending=False)

country_uva_total_dons
```

Let's also see which countries are most frequently giving to UVA regardless of the sum of their gifts

```python
country_uva_count_dons = uva.groupby('Country of Giftor')['Foreign Gift Amount'].count().sort_values(ascending=False)

country_uva_count_dons
```

Lets repeat the same processes above for VCU, a big public school that is in our same state

```python
vcu = df[df['Institution Name'] == 'Virginia Commonwealth University']

country_vcu_total_dons = vcu.groupby('Country of Giftor')['Foreign Gift Amount'].sum().sort_values(ascending=False)

country_vcu_count_dons = vcu.groupby('Country of Giftor')['Foreign Gift Amount'].count().sort_values(ascending=False)

country_vcu_total_dons.head(10)
```

```python
country_vcu_count_dons.head(10)
```
